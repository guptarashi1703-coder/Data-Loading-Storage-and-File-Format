**Load StudentsPerformance.csv**

```python
import pandas as pd
df = pd.read_csv("StudentsPerformance.csv")
```

**Display shape, columns, data types and missing values**

```python
print("Shape:")
print(df.shape)
print("\nColumns:")
print(df.columns)
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
```

**Save it as Excel, JSON, and Parquet**

```python
df.to_excel("StudentsPerformance.xlsx", index=False)
df.to_json("StudentsPerformance.json", orient="records")
df.to_parquet("StudentsPerformance.parquet", index=False)
```

**Reload all files**

```python
excel_df = pd.read_excel("StudentsPerformance.xlsx")
json_df = pd.read_json("StudentsPerformance.json")
parquet_df = pd.read_parquet("StudentsPerformance.parquet")
print("\nFiles reloaded successfully.")
```

**Compare file sizes**

```python
import os
print("\nFile Sizes:")
files = [
    "StudentsPerformance.csv",
    "StudentsPerformance.xlsx",
    "StudentsPerformance.json",
    "StudentsPerformance.parquet"
]
for file in files:
    size = os.path.getsize(file)
    print(file, ":", size, "bytes")
```

**Find the average math, reading, and writing scores**

```python
print("\nAverage Scores:")
print("Math:", df["math score"].mean())
print("Reading:", df["reading score"].mean())
print("Writing:", df["writing score"].mean())
```

**Filter students with math score greater than 80**

```python
filtered_df = df[df["math score"] > 80]
print("\nStudents with Math Score > 80:")
print(filtered_df)
```

**Save the filtered records as a new CSV file**

```python
filtered_df.to_csv("StudentsMathAbove80.csv", index=False)
print("\nFiltered data saved successfully.")
```
